```python
# ==================== PARTE 1 — Criar os Dados ====================
import pandas as pd

dados = pd.DataFrame({
    "horas": [1, 2, 3, 4, 5, 6, 7, 8],
    "aprovado": [0, 0, 0, 1, 0, 1, 1, 1]
})

display(dados)

# ==================== PARTE 2 — Treinar o Modelo ====================
from sklearn.linear_model import LogisticRegression

X = dados[["horas"]]
y = dados["aprovado"]

modelo = LogisticRegression()
modelo.fit(X, y)

print("b0:", modelo.intercept_[0])
print("b1:", modelo.coef_[0][0])

# ==================== PARTE 3 — Calcular as Probabilidades ====================
novos = pd.DataFrame({"horas": [3, 5, 7]})

print("Probabilidades de aprovação:", modelo.predict_proba(novos)[:, 1])
print("Classificações:", modelo.predict(novos))

# ==================== PARTE 4 — Construir o Gráfico ====================
import matplotlib.pyplot as plt

x = pd.DataFrame({"horas": range(1, 9)})
p = modelo.predict_proba(x)[:, 1]

plt.figure(figsize=(8, 5))
plt.scatter(dados["horas"], dados["aprovado"], color="blue", label="Dados reais", s=80)
plt.plot(x["horas"], p, color="red", linewidth=2, label="Probabilidade estimada")
plt.xlabel("Horas de estudo")
plt.ylabel("Probabilidade de aprovação")
plt.title("Regressão Logística - Aprovação x Horas de Estudo")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
